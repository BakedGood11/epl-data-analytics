### Sets
Similar to lists and tuples but **have no duplicates** \
Cannot be accessed via index because it is unordered

```python
daily_pints = {1,2,3,4,5,6,7,8,9,0}
```

```python
type(daily_pints)
```

```python
daily_pints_log = {1,2,3,1,2,5,6,2,3,1,2,2}

print(daily_pints_log)
```

```python
print(daily_pints | daily_pints_log) # Shows the combined sets
```

```python
print(daily_pints & daily_pints_log) # Shows only matches
```

```python
print(daily_pints - daily_pints_log) # Shows what doesn't match
```

```python
print(daily_pints ^ daily_pints_log) # Shows which values are in one set but not in the other
```

# For Loops

```python
integers = [1,2,3,4,5]

for number in integers: #'number' is a temporary variable for this loop, 'integers' is the list in question
    print(number)
```

```python
for number in integers:
        print('yep!')
```

```python
for jelly in integers:
    print(jelly + jelly)
```

```python
ice_cream_dict = {'name': 'Michael D', 'weekly intake': 5, 'favorite ice creams': ['Chocolate', 'Vanilla']}
```

```python
for cream in ice_cream_dict.values():
    print(cream)
```

```python
for key, value in ice_cream_dict.items():
    print(key, '->', value)
```

# Nested For Loop

```python
flavors = ['Vanilla', 'Chocolate', 'Cookie Dough']
toppings = ['Hot Fudge', 'Oreos', 'Marshmallows']
```

```python
for one in flavors:
    for two in toppings:
        print(one, 'topped with', two)
```

# While Loops

```python
number = 0 #Variable we start with

while number < 5: # Code will loop around this block while the condition is met
    print(number)
    number = number + 1
```

```python
number = 0 

while number < 5: 
    print(number)
    if number == 3:
        break     # Breaks the loop before the set condition is met
    number = number + 1
```

```python
number = 0 

while number < 5:
    print(number)
    if number == 6: # Try changing to 3
        break  
    number = number + 1
else:
    print('No longer < 5') # Only works with the "while" loop in this scenario, 
                           # breaking out at ==6 will nullify the else statement
    
```

#### Continue

```python
number = 0 

while number < 5:
    number = number + 1
    if number == 3:
        continue  # Will skil "print" at 3, and reset loop continuing to 4
    print(number)
else:
    print('No longer < 5')
```

# Functions

```python
def function():
    print("My First Function")

function()
```

```python
def first_func():
    print('We did it!')
```

```python
first_func()
```

```python
def number_squared(number):
    print(number**2)
```

```python
number_squared(8)
```

```python
def number_squared_cust(number,power):
    print(number**power) # Needs users to specify both arguments
```

```python
number_squared_cust(5,3)
```

```python
def number_args(*number):
    print(number[0]*number[1]) # [0] and [1] denote index positions
```

```python
number_args(5,4,4,5,6) # with [0] and [1] as 5 and 4
```

```python
args_tuple = (5,4,4,5,6) # we can insert tuples into the function

number_args(*args_tuple)
```

```python
def number_squared_cust(number,power):
    print(number**power)

number_squared_cust(power = 5,number = 3) # By specifying which is which via keyword 
                                          # arguments we get more control 
                                          # over the arguments
```

```python
# We can use arbitrary keyword arguments when we don't know how many 
# arguments our code will end up having
def number_kwarg(**number):             
    print('My number is: ' + number['integer'] + '. My other number: ' + number['integer2'])  
    
    
```

```python
number_kwarg(integer = '2309', integer2 = '546')
```
